```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("../data/student_scores.csv")

data
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
plt.scatter(data["Age"], data["Score"])

plt.xlabel("Age")
plt.ylabel("Score")
plt.title("Age vs Score")

plt.show()
```

```python
data.groupby("AI")["Score"].mean()
```

```python
data.groupby("ML")["Score"].mean().plot(kind="bar")

plt.xlabel("ML Knowledge (0 = No, 1 = Yes)")
plt.ylabel("Average Score")
plt.title("Average Score by ML Knowledge")

plt.show()
```

```python
data.corr()
```

```python
sns.heatmap(data.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```
